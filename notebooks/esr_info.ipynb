# Commodities

```python
import requests
import pandas as pd
import os

# USDA API Key
USDA_API_KEY = os.getenv("USDA_API_KEY")

# Base URL
BASE_URL = "https://api.fas.usda.gov"

# Parameters for request
params = {"api_key": USDA_API_KEY}
ENDPOINT = f"/api/esr/commodities"

response = requests.get(BASE_URL + ENDPOINT, params=params)

if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame(data)
    commodities = pd.concat(
        [
            df[df["commodityName"].str.contains("corn", case=False)],
            df[df["commodityName"].str.contains("soybean", case=False)],
            df[df["commodityName"].str.contains("wheat", case=False)]
        ],
        ignore_index=True,
    )
else:
    print(f"Error: {response.status_code}, {response.text}")

commodities
```

# Countries

```python
ENDPOINT = f"/api/esr/countries"
response = requests.get(BASE_URL + ENDPOINT, params=params)

if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame(data)
    countries = pd.concat(
        [
            df[df["countryDescription"].str.contains("mexico", case=False)],
            df[df["countryDescription"].str.contains("japan", case=False)],
            df[df["countryDescription"].str.contains("korea, republic", case=False)], # South Korea
            df[df["countryDescription"].str.contains("colombia", case=False)],
            df[df["countryDescription"].str.contains("european union", case=False)],
            df[df["countryDescription"].str.contains("taiwan", case=False)],
            df[df["countryDescription"].str.contains("guatemala", case=False)],
            df[df["countryDescription"].str.contains("vietnam", case=False) & ~df["countryDescription"].str.contains("south", case=False)],
            df[df["countryDescription"].str.contains("canada", case=False)],
            df[df["countryDescription"].str.contains("honduras", case=False)],
            df[df["countryDescription"].str.contains("philippines", case=False)],
            df[df["countryDescription"].str.contains("nigeria", case=False)],
            df[df["countryDescription"].str.contains("indonesia", case=False)],
            df[df["countryDescription"].str.contains("thailand", case=False)],
            df[df["countryDescription"].str.contains("china", case=False)],
            df[df["countryDescription"].str.contains("egypt", case=False)],
            df[df["countryDescription"].str.contains("pakistan", case=False)],
            df[df["countryDescription"].str.contains("bangladesh", case=False)]
        ],
        ignore_index=True,
    )
else:
    print(f"Error: {response.status_code}, {response.text}")

countries
```

# Units

```python
unit_ids = [1] # From selected commodities

ENDPOINT = f"/api/esr/unitsOfMeasure"
response = requests.get(BASE_URL + ENDPOINT, params=params)

if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame(data)
    units = df[df["unitId"].isin(unit_ids)]
else:
    print(f"Error: {response.status_code}, {response.text}")

units
```
